##REGRESIÓN LINEAL


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df=pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
x = df[["horas"]]
y = df[["nota"]]
print(x)
```

```python
modelo=LinearRegression()
modelo.fit(x, y)
```

```python
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
prediccion = modelo.predict ([[6]])
print("Prediccion: ", prediccion)
```

```python
# 1. Graficar los datos originales (puntos azules)
plt.scatter(x, y, color='blue', label='Datos reales')

# 2. Graficar la línea de regresión (línea roja)
plt.plot(x, modelo.predict(x), color='red', label='Línea de regresión')

# 3. Graficar el nuevo punto predicho (estrella verde)
valor_predicho = prediccion[0][0]
plt.scatter(6, valor_predicho, color='green', marker='*', s=150, label=f'Predicción: {valor_predicho:.2f}')

plt.title('Regresión Lineal: Horas vs Notas')
plt.xlabel('Horas')
plt.ylabel('Nota')
plt.legend()
plt.grid(True)
plt.show()
```
